## Loaded `train.csv`, displayed first rows, checked data types and summary statistics.


```python
import pandas as pd
df = pd.read_csv('train.csv') 
print(df.head())
print(df.info())
print(df.describe())
```

## Checked for missing values by printing the sum of null values in each column.


```python
print(df.isnull().sum())
```

## Printed the normalized value counts of the 'Survived' column to check the distribution.


```python
print(df['Survived'].value_counts(normalize=True))
```

## Plotted survival count by gender using a countplot and added a title.


```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Survived', hue='Sex', data=df)
plt.title('Survival Count by Gender')
plt.show()
```

## Plotted survival count by passenger class using a countplot and added a title.


```python
sns.countplot(x='Survived', hue='Pclass', data=df)
plt.title('Survival Count by Passenger Class')
plt.show()
```

## Plotted the age distribution with a histogram and KDE (Kernel Density Estimation).


```python
sns.histplot(df['Age'], kde=True)
plt.title('Age Distribution')
plt.show()
```

## Plotted the fare distribution by passenger class using a boxplot and added a title.


```python
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title('Fare Distribution by Class')
plt.show()
```
